##  Initial Observations

### Dataset Overview
- The dataset contains **99,441 order-level records and 15 variables** covering order details, customer identifiers, payment information, delivery timestamps, and review scores.
- `order_id` and `customer_id` are unique in this prepared dataset, indicating that each row represents a distinct order record.

### Data Quality
- No completely duplicated rows were identified.
- Missing values are present in several columns, with the highest proportion found in `order_delivered_customer_date` (approximately 2.98%).
- Missing delivery-related dates may be associated with orders that were not successfully delivered and will require further investigation before treatment.
- `n_items`, `price`, and `freight` each have approximately 0.78% missing values, while `review_score` has approximately 0.77% missing values.

### Order and Payment Overview
- The majority of orders have a `delivered` status, accounting for approximately 97% of the dataset.
- `credit_card` is the most frequently recorded payment method, followed by `boleto`.
- Review scores range from 1 to 5, providing an opportunity to study customer satisfaction in relation to order and delivery characteristics.

### Data Preparation Requirements
- The date-related columns are currently stored as non-datetime values and will require appropriate datetime conversion during preprocessing.
- Missing values will be investigated based on their business context rather than being removed or filled blindly.
- Numerical variables such as `price`, `freight`, and `payment` show wide ranges and will require further distribution and outlier analysis.

### Next Steps
- Investigate the causes and patterns of missing values.
- Convert and validate date/time columns.
- Perform data cleaning and preprocessing.
- Create useful analytical features such as delivery duration and delivery delay.
- Continue with purchase, payment, delivery, and customer satisfaction analysis.
