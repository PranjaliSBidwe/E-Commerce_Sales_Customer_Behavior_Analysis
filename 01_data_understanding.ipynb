# 3. Data Cleaning & Preprocessing

## 3.1 Create a Clean Dataset

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Clean dataset shape:", df_clean.shape)
```

```python
before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Rows removed:", before - after)
```

```python
print("Unique order IDs:", df_clean["order_id"].nunique())
print("Total rows:", len(df_clean))
```

## 3.2 Standardize Categorical Values

```python
categorical_columns = [
    "order_status",
    "payment_type"
]

for column in categorical_columns:
    df_clean[column] = df_clean[column].str.strip().str.lower()
```

```python
for column in categorical_columns:
    print(f"\n{column}:")
    print(df_clean[column].value_counts(dropna=False))
```

## 3.3 Convert Date Columns

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    df_clean[column] = pd.to_datetime(
        df_clean[column],
        errors="coerce"
    )
```
