# 3. Data Cleaning & Preprocessing

## 3.1 Create a Clean Dataset

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Clean dataset shape:", df_clean.shape)
```

```python
before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Rows removed:", before - after)
```

```python
print("Unique order IDs:", df_clean["order_id"].nunique())
print("Total rows:", len(df_clean))
```

## 3.2 Standardize Categorical Values

```python
categorical_columns = [
    "order_status",
    "payment_type"
]

for column in categorical_columns:
    df_clean[column] = df_clean[column].str.strip().str.lower()
```

```python
for column in categorical_columns:
    print(f"\n{column}:")
    print(df_clean[column].value_counts(dropna=False))
```

## 3.3 Convert Date Columns

# 4. Feature Engineering

```python
df_clean.columns.tolist()
```

## 4.1 Delivery Duration

```python
df_clean["delivery_days"] = (
    df_clean["order_delivered_customer_date"]
    - df_clean["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
df_clean["delivery_days"].describe()
```

## 4.2 Expected Delivery Duration

```python
df_clean["expected_delivery_days"] = (
    df_clean["order_estimated_delivery_date"]
    - df_clean["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
df_clean["expected_delivery_days"].describe()
```

## 4.3 Delivery Delay

```python
df_clean["delivery_delay_days"] = (
    df_clean["order_delivered_customer_date"]
    - df_clean["order_estimated_delivery_date"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
df_clean["delivery_delay_days"].describe()
```

## 4.4 Delivery Performance Category

```python
df_clean["delivery_performance"] = np.select(
    [
        df_clean["delivery_delay_days"] < 0,
        df_clean["delivery_delay_days"] == 0,
        df_clean["delivery_delay_days"] > 0
    ],
    [
        "Early",
        "On Time",
        "Late"
    ],
    default="Not Delivered"
)
```

```python
df_clean["delivery_performance"].value_counts(dropna=False)
```

## 4.5 Purchase Date Features

```python
df_clean["purchase_year"] = (
    df_clean["order_purchase_timestamp"].dt.year
)
```

```python
df_clean["purchase_month"] = (
    df_clean["order_purchase_timestamp"].dt.month
)
```

```python
df_clean["purchase_month_name"] = (
    df_clean["order_purchase_timestamp"].dt.month_name()
)
```

```python
df_clean["purchase_day"] = (
    df_clean["order_purchase_timestamp"].dt.day
)
```

```python
df_clean["purchase_weekday"] = (
    df_clean["order_purchase_timestamp"].dt.day_name()
)
```

```python
df_clean["purchase_hour"] = (
    df_clean["order_purchase_timestamp"].dt.hour
)
```
